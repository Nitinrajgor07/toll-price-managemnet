## 📈 Step 3 — Final Result Charts

````python
# Chart: All 4 phases summary
fig, axes = plt.subplots(2, 2, figsize=(14, 8))
fig.suptitle('Smart Toll Pricing System — Complete Project Overview', 
             fontsize=14, fontweight='bold', y=1.02)

# 1. Hourly Traffic Pattern
df_raw = pd.read_csv('data/processed_traffic.csv')
hourly = df_raw.groupby('hour')['traffic_volume'].mean()
axes[0,0].bar(hourly.index, hourly.values, color='#1a237e', alpha=0.8, width=0.7)
axes[0,0].axvspan(7,9,   alpha=0.2, color='red')
axes[0,0].axvspan(16,18, alpha=0.2, color='orange')
axes[0,0].set_title('Phase 1: Traffic Pattern Analysis', fontweight='bold')
axes[0,0].set_xlabel('Hour'); axes[0,0].set_ylabel('Avg Volume')
axes[0,0].grid(axis='y', alpha=0.3)

# 2. Congestion Distribution
COLORS = ['#28a745','#ffc107','#dc3545']
dist   = df_raw['congestion_level'].value_counts().sort_index()
axes[0,1].pie(dist.values, labels=['Low','Medium','High'],
              colors=COLORS, autopct='%1.1f%%',
              wedgeprops=dict(width=0.6))
axes[0,1].set_title('Phase 2: Congestion Distribution', fontweight='bold')

# 3. Model Metrics
metric_names  = ['Accuracy','Precision','Recall','F1 Score']
metric_values = [
    metrics['accuracy']*100,
    metrics['precision']*100,
    metrics['recall']*100,
    metrics['f1_score']*100
]
bars = axes[1,0].bar(metric_names, metric_values,
                     color=['#1a237e','#28a745','#ffc107','#dc3545'], width=0.5)
axes[1,0].set_ylim(0,110)
axes[1,0].axhline(y=80, color='red', linestyle='--', alpha=0.5)
axes[1,0].set_title('Phase 2: Model Performance', fontweight='bold')
axes[1,0].set_ylabel('Score (%)')
axes[1,0].grid(axis='y', alpha=0.3)
for bar,val in zip(bars,metric_values):
    axes[1,0].text(bar.get_x()+bar.get_width()/2, val+1,
                   f'{val:.1f}%', ha='center', fontweight='bold', fontsize=9)

# 4. Dynamic Pricing
scenarios = ['Mon\nRush','Fri\nEvening','Rainy\nDay','Normal\nDay','Sun\nNight']
tolls     = [180, 144, 96, 80, 50]
colors    = ['#dc3545','#dc3545','#ffc107','#ffc107','#28a745']
bars2 = axes[1,1].bar(scenarios, tolls, color=colors, width=0.5)
for bar,val in zip(bars2,tolls):
    axes[1,1].text(bar.get_x()+bar.get_width()/2, val+2,
                   f'₹{val}', ha='center', fontweight='bold', fontsize=10)
axes[1,1].set_title('Phase 3: Dynamic Toll Pricing', fontweight='bold')
axes[1,1].set_ylabel('Toll Price (₹)')
axes[1,1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.savefig('static/charts/project_overview.png', dpi=120, bbox_inches='tight')
plt.show()
print('✅ Complete project overview chart saved!')
````
